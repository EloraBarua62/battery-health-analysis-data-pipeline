```python
# Connection configuration
connectionString = "Endpoint=sb://your-name.servicebus.windows.net/;..."
conf = {"eventhubs.connectionString": sc._jvm.org.apache.spark.eventhubs.EventHubsUtils.encrypt(connectionString)}

# Start the stream
df = spark.readStream \
    .format("eventhubs") \
    .options(**conf) \
    .load()

# Convert binary body to string and display
from pyspark.sql.functions import col
display(df.select(col("body").cast("string")))
```
